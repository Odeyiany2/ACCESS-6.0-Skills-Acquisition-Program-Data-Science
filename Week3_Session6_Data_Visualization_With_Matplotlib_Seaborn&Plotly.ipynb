# 1.)BAR CHARTS - Comparing Categories

**When to use:** Comparing values across different categories(countries, products, regions)

**Business Question:** Which countries have the highest stock index values?

````python
df.head()
````

````python
#Get the top 10 Countries by index_value
top_10 = df.sort_values('Index_Value', ascending = False).head(10)

#Method 1 - Matplotlib
plt.figure(figsize=(12, 6)) #Set the chart size. Width = 12 inches and height = 6 inches

plt.bar( x= top_10['Country'],
         height= top_10['Index_Value'],
         color = 'skyblue',
         edgecolor = 'black',
         linewidth = 2
        )
plt.title('Top 10 Countries by Stock Index Value', fontsize=16, fontweight='bold')
plt.xlabel('Country', fontsize=12)
plt.ylabel('Index Value', fontsize=12)
plt.xticks(rotation=45)

#Add value labels on top of bars
for index, value in enumerate(top_10['Index_Value']):
    plt.text(x=index, y=value, s=f'{value}', ha='center', va='bottom', fontsize=10)

plt.tight_layout()
plt.show()

````

````python
#Seaborn
plt.figure(figsize =(12,6))
sns.barplot(data =top_10,
            x ='Country',
            y ='Index_Value',
            palette = 'rocket'
            )
plt.title('Top 10 Countries by Stock Index Value', fontsize=16, fontweight='bold')
plt.xlabel('Country', fontsize=12)
plt.ylabel('Index Value', fontsize=12)
plt.xticks(rotation=45)
plt.tight_layout()
plt.show()
````

````python
#Get the top 10 coutries by index value
top_countries = df.nlargest(10, 'Index_Value')[['Country', 'Index_Value']]


plt.figure(figsize=(12, 6))

# Create bar plot - Seaborn style
sns.barplot(data= top_countries,
            x='Country',
            y='Index_Value',
            palette='rocket')  # Color palette

plt.title('Top 10 Countries by Stock Index Value (Seaborn)',
          fontsize=16,
          fontweight='bold')
plt.xlabel('Country', fontsize=12)
plt.ylabel('Index Value', fontsize=12)
plt.xticks(rotation=45, ha='right')
plt.tight_layout()
plt.show()
````

# **2.)HISTOGRAMS - Understanding Distributions**

**When to use:** Understanding how values are distributed, finding patterns in continuous data.

**Business Question:** How are inflation rates distributed across countries?

````python
# What is a histogram?
# It divides data into "bins" (ranges) and counts how many values fall in each bin
# Example: If we have inflation rates from 0% to 10%,
# bins might be: 0-2%, 2-4%, 4-6%, etc.


plt.figure(figsize=(10, 6))

# Histogram with smooth curve overlay
sns.histplot(data=df,
             x='Inflation_Rate_Percent',
             bins=20,
             kde=True,  # Add smooth density curve
             color='skyblue',
             edgecolor='black')

plt.title('Inflation Rate Distribution with Density Curve',
          fontsize=16,
          fontweight='bold')
plt.xlabel('Inflation Rate (%)', fontsize=12)
plt.ylabel('Frequency', fontsize=12)
plt.legend()
plt.tight_layout()
plt.show()
````

# **3.)BOXPLOTS - Spotting Outliers and Comparing Groups**
**When to use:** Comparing distributions across categories, identifying outliers.

**Business Question:** How do Interest rates vary across different credit rating categories?

````python
# What does a boxplot show?
# - Box: Middle 50% of data (25th to 75th percentile)
# - Line in box: Median (50th percentile)
# - Whiskers: Extend to 1.5 × IQR (Interquartile Range)
# - Dots beyond whiskers: Outliers

plt.figure(figsize=(12, 6))

# Boxplot comparing interest rates across credit ratings
sns.boxplot(data=df,
            x='Credit_Rating',
            y='Interest_Rate_Percent',
            palette='Set2')

plt.title('Interest Rates by Credit Rating Category',
          fontsize=16,
          fontweight='bold')
plt.xlabel('Credit Rating', fontsize=12)
plt.ylabel('Interest Rate (%)', fontsize=12)
plt.grid(axis='y', alpha=0.3)
plt.tight_layout()
plt.show()

````

````python
# Violin plot (Boxplot + Distribution)

plt.figure(figsize=(12, 6))

# Violin plot shows distribution shape + boxplot statistics
sns.violinplot(data=df,
               x='Credit_Rating',
               y='Unemployment_Rate_Percent',
               palette='muted')

plt.title('Unemployment Rate Distribution by Credit Rating (Violin Plot)',
          fontsize=16,
          fontweight='bold')
plt.xlabel('Credit Rating', fontsize=12)
plt.ylabel('Unemployment Rate (%)', fontsize=12)
plt.tight_layout()
plt.show()
````

# **4.)SCATTER PLOTS - Finding Relationships**

**When to use:** Exploring relationships between two continuous variables.

**Business Question:** Is there a relationship between GDP growth and stock market performance?


````python
# Seaborn with regression line

plt.figure(figsize=(10, 6))

# Scatter plot with trend line
sns.regplot(data=df,
            x='GDP_Growth_Rate_Percent',
            y='Index_Value',
            scatter_kws={'alpha':0.6, 's':100, 'edgecolor':'black'},
            line_kws={'color':'red', 'linewidth':2}
)

plt.title('GDP Growth vs Index Value with Trend Line',
          fontsize=16,
          fontweight='bold')
plt.xlabel('GDP Growth Rate (%)', fontsize=12)
plt.ylabel('Stock Index Value', fontsize=12)
plt.grid(True, alpha=0.3)
plt.tight_layout()
plt.show()

# Calculate correlation
correlation = df['GDP_Growth_Rate_Percent'].corr(df['Index_Value'])
print(f"correlation coefficient: {correlation}")
````

**Understanding correlation:**
```
+1: Perfect positive relationship (as X increases, Y increases)
0: No linear relationship
-1: Perfect negative relationship (as X increases, Y decreases)
|r| > 0.7: Strong
0.4 < |r| < 0.7: Moderate
|r| < 0.4: Weak```

````python
# Pairplot for multiple relationships

# Select numeric columns for comparison
numeric_cols = ['GDP_Growth_Rate_Percent', 'Inflation_Rate_Percent',
                'Interest_Rate_Percent', 'Unemployment_Rate_Percent',
                'Index_Value']

# Create pairplot
pairplot_fig = sns.pairplot(df[numeric_cols],
                            diag_kind='kde',  # Diagonal shows distribution
                            plot_kws={'alpha':0.6, 's':50},
                            diag_kws={'alpha':0.7})

pairplot_fig.fig.suptitle('Pairwise Relationships in Economic Indicators',
                          y=1.02,
                          fontsize=16,
                          fontweight='bold')
plt.tight_layout()
plt.show()
````

**Pairplot explanation**: Shows scatter plots for every combination of variables. Diagonal shows the distribution of each variable. This is your "at-a-glance" correlation checker!

# **5.)TIME SERIES PLOTS- Tracking Changes Over Time**
**When to use**: Showing trends, seasonality, and patterns over time.
**Business Question**: How do market value change over time, and are there patterns?

````python

# First, convert Date column to datetime
df['Date'] = pd.to_datetime(df['Date'])

# Sort by date
df_time = df.sort_values('Date').reset_index(drop=True)

print("Date Range in Dataset:")
print(f"   Start: {df_time['Date'].min().strftime('%Y-%m-%d')}")
print(f"   End: {df_time['Date'].max().strftime('%Y-%m-%d')}")
print(f"   Total Days: {(df_time['Date'].max() - df_time['Date'].min()).days}")
````

````python
# Basic time series line plot

# Select a few countries to visualize (change based on your data)
countries_sample = df_time['Country'].unique()[:5]  # First 5 countries

fig, ax = plt.subplots(figsize=(14, 6))

for country in countries_sample:
    country_data = df_time[df_time['Country'] == country]
    ax.plot(country_data['Date'],
            country_data['Index_Value'],
            marker='o',
            linewidth=2,
            label=country,
            alpha=0.8)

ax.set_title('Stock Index Value Over Time by Country',
             fontsize=16,
             fontweight='bold')
ax.set_xlabel('Date', fontsize=12)
ax.set_ylabel('Index Value', fontsize=12)
ax.legend(loc='best', frameon=True, shadow=True, ncol=2)
ax.grid(True, alpha=0.3, linestyle='--')

# Format dates nicely
import matplotlib.dates as mdates
ax.xaxis.set_major_formatter(mdates.DateFormatter('%b %d'))
plt.xticks(rotation=45)

plt.tight_layout()
plt.show()
````

#The .plot() Method - Pandas' Secret Weapon

Pandas DataFrames have a built-in .plot() method that's quick and convenient for exploratory analysis.

````python
# Basic syntax examples

# 1. Line plot (default)
df.groupby('Date')['Index_Value'].mean().plot(figsize=(12, 4),
                                               title='Average Index Value Over Time',
                                               ylabel='Index Value',
                                               grid=True,
                                               linewidth=2,
                                               color='purple')
plt.tight_layout()
plt.show()

# 2. Bar plot
df.groupby('Credit_Rating')['Interest_Rate_Percent'].mean().plot(
    kind='bar',
    figsize=(10, 5),
    color='teal',
    title='Average Interest Rate by Credit Rating',
    edgecolor='black',
    alpha=0.7
)
plt.ylabel('Average Interest Rate (%)')
plt.xlabel('Credit Rating')
plt.xticks(rotation=45)
plt.grid(axis='y', alpha=0.3)
plt.tight_layout()
plt.show()

# 3. Histogram
df['GDP_Growth_Rate_Percent'].plot(
    kind='hist',
    bins=20,
    figsize=(10, 5),
    title='GDP Growth Rate Distribution',
    edgecolor='black',
    color='green',
    alpha=0.7
)
plt.xlabel('GDP Growth Rate (%)')
plt.ylabel('Frequency')
plt.grid(axis='y', alpha=0.3)
plt.tight_layout()
plt.show()

# 4. Scatter plot
df.plot(
    kind='scatter',
    x='Inflation_Rate_Percent',
    y='Unemployment_Rate_Percent',
    c='Interest_Rate_Percent',  # Color by another column
    cmap='viridis',
    s=100,
    figsize=(10, 6),
    title='Inflation vs Unemployment (colored by Interest Rate)',
    alpha=0.6,
    edgecolors='black'
)
plt.xlabel('Inflation Rate (%)')
plt.ylabel('Unemployment Rate (%)')
plt.grid(True, alpha=0.3)
plt.tight_layout()
plt.show()

# 5. Box plot
df[['GDP_Growth_Rate_Percent', 'Inflation_Rate_Percent',
    'Interest_Rate_Percent', 'Unemployment_Rate_Percent']].plot(
    kind='box',
    figsize=(12, 6),
    title='Economic Indicators Distribution Comparison',
    grid=True
)
plt.ylabel('Percentage (%)')
plt.xticks(rotation=45)
plt.tight_layout()
plt.show()

# 6. Area plot
df.groupby('Date')[['GDP_Growth_Rate_Percent',
                    'Inflation_Rate_Percent']].mean().plot(
    kind='area',
    figsize=(14, 6),
    title='GDP Growth and Inflation Over Time',
    alpha=0.5,
    stacked=False
)
plt.ylabel('Rate (%)')
plt.xlabel('Date')
plt.legend(loc='upper left')
plt.grid(True, alpha=0.3)
plt.tight_layout()
plt.show()
````
