### **Reading and Inspecting Methods**
1. head

head displays the first few rows of the dataset.
It helps you quickly see how the data looks — the structure, sample entries, and whether columns loaded correctly.

2. tail

tail shows the last few rows of the dataset.
It is useful for checking the end of the file, especially when the data may have inconsistencies or formatting issues at the bottom.

3. columns

columns lists the names of every column in the dataset.
This helps you understand the features available, confirm correct spelling, and identify which columns you want to work with.

4. info

info provides a full summary of the dataset, including:

number of rows and columns

each column’s data type

whether a column has missing values

memory usage

It is one of the most important inspection tools because it tells you the structure and health of your dataset.

5. shape

shape shows the size of your dataset in the form (number of rows, number of columns).
This helps you understand how large your dataset is and whether it loaded correctly.

6. describe

describe gives statistical summaries for all numerical columns, including:

average

standard deviation

minimum

maximum

percentiles

It helps you see the distribution and characteristics of your numeric data.

7. loc

loc is used to select data by labels — meaning you use actual row names or column names.
It is intuitive when you want to pick data based on column names or apply conditions (e.g., selecting all rows where Category is “Food”).

_**code example: df_loc = df.loc[df["Category"] == "Food", ["Date", "Amount"]]**_

8. iloc

iloc selects data by position — meaning you use row numbers and column numbers.
It works like selecting items in a list using their index, starting from zero.

_**code example: df_iloc = df.iloc[0:5, 0:3]  # first 5 rows, first 3 columns**_

9. Filtering

Filtering is the process of selecting rows based on a condition.
For example, you might filter:

all transactions greater than a certain amount

all records belonging to a specific category

all rows in a date range

Filtering helps narrow down the dataset to what is relevant for analysis.

_**code example: df_filtered = df[df["Amount"] > 1000]**_

10. Aggregation

Aggregation means combining or summarizing data.
Examples include calculating:

total spending per category

average amount per month

number of transactions per account

Aggregation is often done using grouping, and it helps you create insights from raw data by summarizing patterns and trends.

_**code example: df_agg = df.groupby("Category")["Amount"].sum()**_
